```python
n=5
for row in range(2*n-1):
    if row<n:
        spacing = n-(row+1)
        for col in range(spacing):
            print(" ",end = " ")
        for col in range((2*row)+1):
            print("*",end=" ")
        print()
    elif row>=n:
        spacing = row - n 
        for col in range(spacing+1):
            print(" ",end = " ")
        for col in range(((2*n-2)-row) + ((2*n-1)-row)):
            print("*",end = " ")
        print()
```

```python
alpha = {0:"A",1:"B",2:"C",3:"D"}
n = 4

for row in range(n):
    spacing = n - (row+1) 
    for col in range(spacing):
        print(" ",end=" ")
    for col in range(row+1):
        print(alpha[col],end=" ")
    for col in range(row,0,-1):
        print(alpha[col-1],end=" ")
    for col in range(spacing):
        print(" ",end=" ")
    print()
        
```

```python
n = 5
for row in range(2*n):

    if row<n:
        for col in range(n-row):
            print("*",end= " ")

        for col in range(2*row):
            print(" ",end = " ")

        for col in range(n-row):
            print("*",end = " ")
    elif row>=n:
        for col in range((row+1)-n):
            print("*",end = " ")

        for col in range(((2*n)-(row+1))*2):
            print(" ",end = " ")

        for col in range(row-n+1):
            print("*",end = " ")

    print()
```

```python
n = 5
tot_row = 2*n-1
for row in range(tot_row):
    if row < n : 
        for col in range(row+1):
            print("*",end = " ")

        for col in range((tot_row)-(row*2)-1):
            print(" ",end=" ")

        for col in range(row+1):
            print("*",end=" ")

    if row >= n:
        for col in range(tot_row-row):
            print("*",end=" ")

        for col in range((n - (tot_row-row))*2):
            print(" ",end=" ")

        for col in range(tot_row-row):
            print("*",end=" ")
        
    print()
```
