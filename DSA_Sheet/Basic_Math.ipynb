## Reverse Integer

```python
class Solution(object):
    def reverse(self, x):
        """
        :type x: int
        :rtype: int
        """
        if x > 0 and int(str(x)[::-1]) < 2147483647:
            return int(str(x)[::-1])
        elif x < 0 and int("-" + str(x)[::-1][:-1]) > -2147483648:
            return int("-" + str(x)[::-1][:-1])
        else:
            return 0
```

```python
x = 1223
Solution().reverse(x)
```

## Palindrome Number

```python
class Solution(object):
    def isPalindrome(self, x):
        """
        :type x: int
        :rtype: bool
        """
        if str(x) == str(x)[::-1]:
            return True
        else:
            return False
```

```python
x = 121
Solution().isPalindrome(x)
```

## Find GCD of two numbers

```python
class Solution(object):
    def GCD(self,N1,N2):
        while N2 != 0:
            N1,N2 = N2,N1%N2
        return N1
```

```python
N1,N2 = 9,12
Solution().GCD(N1,N2)
```

## Check if a number is Armstrong Number or not

```python
class Solution(object):
    def Armstrong(self,N):
        if N > 0 :
            str_n = str(N)
            sums = 0
            for i in range(len(str_n)):
                sums += int(str_n[i])**len(str_n)
            if sums == N:
                return True
            else:
                return False

        
```

```python
N = 371
Solution().Armstrong(N)
```

## Print all Divisors of a given Number

```python
class Solution(object):
    def Divisor(self,N):
        if N > 0 :
            Divisor = []
            for i in range(1,N+1):
                if N%i == 0:
                    Divisor.append(i)
        return Divisor
```

```python
N = 36
Solution().Divisor(N)
```

## Prime or not

```python
class Solution(object):
    def isPrime(self,N):
        if N > 0:
            if N <= 3:
                return("Prime")
            elif N > 3:
                for i in range(2,N-1):
                    if N % i == 0:
                        return("Not Prime")
                        break

                return("Prime")
```

```python
N = 123
Solution().isPrime(N)
```
