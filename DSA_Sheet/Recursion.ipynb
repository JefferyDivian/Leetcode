# Recursion Notes

## What is Recursion?

A function that calls itself with a smaller version of the problem, until it hits a stopping condition.

> Think of it like Russian dolls — open one, find a smaller one inside, keep going until you find the one that doesn't open. Then you close them all back up on the way out.

---

## The 2 Rules (Never Break These)

1. **Always have a base case** — the condition where the function stops calling itself
2. **Always move toward the base case** — every recursive call must make the problem smaller

Break either rule → infinite loop → `RecursionError: maximum recursion depth exceeded`

---

## Anatomy of a Recursive Function

```python
def factorial(n):
    if n == 0:               # Rule 1: base case — STOP here
        return 1
    return n * factorial(n - 1)  # Rule 2: moving toward 0
```

---

## How Factorial Works — Step by Step

`factorial(5)` doesn't compute straight away. It builds a stack of waiting calls first.

### Phase 1 — Going In (stack builds up)

```
factorial(5)  → needs factorial(4) first... waiting
  factorial(4)  → needs factorial(3) first... waiting
    factorial(3)  → needs factorial(2) first... waiting
      factorial(2)  → needs factorial(1) first... waiting
        factorial(1)  → needs factorial(0) first... waiting
          factorial(0)  → n == 0, return 1  ✅ BASE CASE
```

### Phase 2 — Coming Back (answers bubble up)

```
factorial(0)  → returns 1
factorial(1)  → returns 1 × 1  = 1
factorial(2)  → returns 2 × 1  = 2
factorial(3)  → returns 3 × 2  = 6
factorial(4)  → returns 4 × 6  = 24
factorial(5)  → returns 5 × 24 = 120  🎉
```

The final answer only appears at the very end, on the way **out** — not on the way in.

---

## The Call Stack

Every recursive call is stored in memory (the **call stack**) until the base case is reached.

- `factorial(5)` → uses **6 stack frames** (n=5 down to n=0)
- Deeper recursion = more memory used
- Stack frames are freed as calls return

```
[ factorial(5) ]  ← top (newest)
[ factorial(4) ]
[ factorial(3) ]
[ factorial(2) ]
[ factorial(1) ]
[ factorial(0) ]  ← bottom (hits base case first)
```

## Another Example — Countdown

```python
def countdown(n):
    if n == 0:          # base case
        print("Go! 🚀")
        return
    print(n)
    countdown(n - 1)    # recursive call

countdown(3)
# Output: 3  2  1  Go! 🚀
```

---

## Common Mistake ❌

```python
def oops(n):
    return oops(n)   # no base case, n never changes → infinite loop
```

Always ask: *"Does each call bring me closer to the base case?"*

---

## When to Use Recursion

Recursion shines when the problem has a **naturally nested or self-similar structure**:

- Factorials and math sequences
- Fibonacci numbers
- Tree traversal (file systems, family trees, DOM)
- Binary search
- Sorting algorithms (merge sort, quicksort)
- Backtracking problems (mazes, puzzles)

---

## Recursion vs Loops

| | Recursion | Loop |
|---|---|---|
| Best for | Trees, nested structures | Simple repetition |
| Memory | Uses call stack (more) | Constant |
| Readability | Often cleaner for complex problems | Simpler for flat tasks |
| Risk | Stack overflow if too deep | Infinite loop if no exit |

---

## Quick Reference

```python
def recursive_fn(problem):
    # 1. Base case first — always
    if problem is small enough:
        return simple_answer

    # 2. Shrink the problem, then recurse
    smaller = make_smaller(problem)
    return do_something(recursive_fn(smaller))
```

---

## Key Terms

| Term | Meaning |
|---|---|
| **Base case** | The condition that stops the recursion |
| **Recursive call** | When the function calls itself |
| **Call stack** | Memory storing all the waiting function calls |
| **Stack frame** | One function call's slot in memory |
| **RecursionError** | Python's crash when the stack gets too deep (default limit: 1000) |

## Print Name N times using Recursion
### Problem Description: Given an integer N, write a program to print your name N times.

````python
def recursion_name(name:str , N:int):
    if N==1:
        print(name)
        return

    print(name,end=" ")
    return recursion_name(name,N=N-1)
````

````python
recursion_name(name="JD",N=6)
````

## Print 1 to N using Recursion
### Problem Description: Given an integer N, write a program to print numbers from 1 to N.

````python
def recursion_function(start,N):

    if start>N:
        raise ValueError("Start of the sequence cant be bigger than the end.")
    
    ## base case:
    if start == N:
        print(N)
        return

    ## main function
    print(start)
    return recursion_function(start+1,N) 
````

````python
recursion_function(1,8)
````

## Print N to 1 using Recursion
### Problem Description: Given an integer N, write a program to print numbers from N to 1.

````python
def print_n_to_1(N):
    if N == 1:
        print(N)
        return

    print(N)
    return print_n_to_1(N-1)
````

````python
print_n_to_1(5)
````

## Sum of first N Natural Numbers
### Problem Statement: Given a number ‘N’, find out the sum of the first N natural numbers .

````python
def recursion_sum(N):
    if N == 1:
        return 1

    return N + recursion_sum(N-1)
````

````python
recursion_sum(6)
````

## Factorial of a Number : Iterative and Recursive

#### Problem Statement: Given a number X,  print its factorial. To obtain the factorial of a number, it has to be multiplied by all the whole numbers preceding it. More precisely X! = X*(X-1)*(X-2) … 1. Note: X  is always a positive number. 

````python
def recursion_factorial(N):
    if N == 1:
        return 1

    return N * recursion_factorial(N-1)
````

````python
recursion_factorial(5)
````

## Reverse a given Array
### Problem Statement: You are given an array. The task is to reverse the array and print it.

````python
arr = [5,4,3,2,1]
def reverse(arr):
    return arr[::-1]

print(reverse(arr))
````

## Check if the given String is Palindrome or not
### Problem Statement: Given a string, check if the string is palindrome or not. A string is said to be palindrome if the reverse of the string is the same as the string.

````python
str = "ABCDCBA"

def palindrome_checker(str):
    return str == str[::-1]

palindrome_checker(str)
````

## Print Fibonacci Series up to Nth term
### Problem Statement: Given an integer N. Print the Fibonacci series up to the Nth term.

````python
def fibonacci(N):
    if N <= 1:
        return (N)

    return fibonacci(N-1) + fibonacci(N-2)
````
