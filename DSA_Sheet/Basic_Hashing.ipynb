# Hashing Notes

## What is Hashing?

A technique that takes an input (of any size) and runs it through a **hash function** to produce a fixed-size output — called a **hash**, **hash code**, or **digest**. That output is then used to quickly locate, compare, or verify data.

> Think of it like a coat check. You hand over your coat (the data), get a numbered ticket (the hash), and that ticket tells you exactly which hook to check — no need to search every hook in the building.

---

## The 2 Properties (Never Break These)

1. **Deterministic** — the same input must always produce the same hash
2. **Fast to compute, uniformly spread** — hashing should be quick, and outputs should spread evenly across the available slots to avoid pile-ups

Break either one → collisions everywhere → your hash table degrades into a slow linked list.

---

## Anatomy of a Hash Table Insert

```python
def hash_function(key, table_size):
    return sum(ord(c) for c in key) % table_size   # Rule 2: spread across slots

def insert(table, key, value):
    index = hash_function(key, len(table))          # Rule 1: same key -> same index
    table[index].append((key, value))                # bucket holds (key, value) pairs
```

---

## How Hashing Works — Step by Step

Inserting `"cat"` into a table of size 7 doesn't just "land somewhere." It goes through a conversion first.

### Phase 1 — Turning Data Into a Number

```
"cat"  → get char codes → c=99, a=97, t=116
       → sum them        → 99 + 97 + 116 = 312
       → mod table_size   → 312 % 7 = 4
```

### Phase 2 — Landing In the Bucket

```
index 4 → table[4] → append ("cat", value)  ✅ STORED
```

### Phase 3 — Looking It Up Later

```
lookup("cat") → hash_function("cat", 7) → 4 (same as before!)
             → check table[4] → find "cat" → return value
```

The hash is only ever *computed*, never stored and searched from scratch — that's what makes lookup fast.

---

## The Hash Table

Every key gets mapped to a **bucket** (an index in an array) based on its hash.

- A table of size 7 → has **7 buckets**, indexes 0 through 6
- More items than buckets = more **collisions** (two keys landing on the same index)
- Buckets typically hold small lists, so collisions don't lose data — they just slow things down

```
[ bucket 0 ]
[ bucket 1 ]  ← "dog" lives here
[ bucket 2 ]
[ bucket 3 ]
[ bucket 4 ]  ← "cat" lives here
[ bucket 5 ]
[ bucket 6 ]
```

---

## Handling Collisions

```python
# Two different keys can hash to the same index.
hash_function("cat", 7)  # → 4
hash_function("fig", 7)  # → 4  (collision!)

# Solution: chaining — store a list at each bucket
table[4] = [("cat", "meow"), ("fig", "fruit")]
```

Two common strategies:

- **Chaining** — each bucket holds a small list; collisions just add to the list
- **Open addressing** — if a slot is taken, probe forward to the next open slot

---

## Common Mistake ❌

```python
def bad_hash(key, table_size):
    return 1   # every key maps to the same bucket → not a hash function, just a dumping ground
```

Always ask: *"Will different inputs actually spread out across the table, or will they all crowd into one bucket?"*

---

## When to Use Hashing

Hashing shines when you need **fast lookup, insertion, or duplicate-checking**:

- Dictionaries / hash maps (`dict` in Python, `HashMap` in Java)
- Caching (mapping a request to a cached response)
- Checking for duplicates in a large dataset
- Password storage (hashing, never storing plain text)
- Data integrity checks (checksums, file verification)
- Database indexing

---

## Hashing vs Searching a List

| | Hashing | Linear Search |
|---|---|---|
| Best for | Fast lookup by key | Small or unsorted data |
| Speed | O(1) average | O(n) — checks every item |
| Memory | Extra space for the table | Constant, no extra structure |
| Risk | Poor hash function → collisions → slow | Always slow, but predictable |

## Quick Reference

```python
def hash_function(key, table_size):
    # 1. Convert the key into a number
    raw_value = convert_to_number(key)

    # 2. Compress it to fit inside the table
    return raw_value % table_size
```

## Key Terms

| Term | Meaning |
|---|---|
| **Hash function** | The formula that converts input into a fixed-size number |
| **Hash / digest** | The output of a hash function |
| **Hash table** | The array of buckets that stores key-value pairs by hash |
| **Bucket** | One slot in the hash table, found via its index |
| **Collision** | When two different keys hash to the same index |
| **Chaining** | Collision strategy: store multiple items per bucket as a list |
| **Load factor** | Ratio of items stored to number of buckets — high load factor = more collisions |

## Count frequency of each element in the array
### Problem Statement: Given an array, we have found the number of occurrences of each element in the array.

````python
arr = [10,5,10,15,10,5]
arr1 = [2,2,3,4,4,2]

def counting(arr):
    counter = {}
    for i in arr:
        if i not in counter:
            counter[i] = 1
        else:
            counter[i]+=1

    return counter
````
