## Find the Largest element in an array
### Problem Statement: Given an array, we have to find the largest element in the array.

```python
arr = [2, 5, 1, 3, 0]
```

```python
def larg_ele(arr):
    larg = arr[0]
    for i in range(1,len(arr)):
        if larg<arr[i]:
            larg = arr[i]
    return larg

larg_ele(arr)
```

## Find Second Smallest and Second Largest Element in an array
### Problem Statement: Given an array, find the second smallest and second largest element in the array. Print ‘-1’ in the event that either of them doesn’t exist.

```python
arr =  [1, 2, 4, 7, 7, 5]  

def second(arr):
    if len(arr)>2:
        sort_arr = sorted(set(arr))
        return sort_arr[1],sort_arr[-2]
    else:
        raise ValueError("There is no second highest and second smallest")
```

```python
second(arr)
```

## Check if an Array is Sorted

### Problem Statement: Given an array of size n, write a program to check if the given array is sorted in (ascending / Increasing / Non-decreasing) order or not. If the array is sorted then return True, Else return False.



```python
def check(nums):
    n = len(nums)
        
    count = 0
    for i in range(n):
        if nums[i]>nums[(i+1) % n]:  ## Check the previous number with the next if there is only count or drop between the previous and the current its fine but mroe that it means is not doing good.
            count += 1

    if count<=1:
        return True
        
    return False
```

```python
nums = [3,4,5,1,2]
check(nums)
```
